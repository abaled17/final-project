```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the Pittsburgh Trees dataset
trees = pd.read_csv("city_trees.csv")

# Show the first five rows
trees.head()
```

```python
# Count trees in each neighborhood
tree_counts = trees.groupby("neighborhood").size()

# Sort neighborhoods from most trees to least
tree_counts = tree_counts.sort_values(ascending=False)

# Display the top 10 neighborhoods
tree_counts.head(10)
```

```python
# Remove stump records
filtered_trees = trees[
    trees["common_name"].fillna("").str.lower() != "stump"
]

# Recalculate the tree counts
tree_counts = filtered_trees.groupby("neighborhood").size()
tree_counts = tree_counts.sort_values(ascending=False)

# Show the updated top 10
print(tree_counts.head(10))

# Create a bar chart
tree_counts.head(10).sort_values().plot(
    kind="barh",
    figsize=(10, 6),
    color="forestgreen"
)

plt.title("Top 10 Pittsburgh Neighborhoods by Recorded Trees")
plt.xlabel("Number of Tree Records")
plt.ylabel("Neighborhood")
plt.tight_layout()
plt.show()
```


### Research Question
Which Pittsburgh neighborhood has the most city-managed trees?

### Dataset
I used the City of Pittsburgh Trees dataset from the Western Pennsylvania Regional Data Center (WPRDC).

### Method
I used Python and pandas to group the tree records by neighborhood and count them. I excluded records labeled "Stump" and created a bar chart showing the top 10 neighborhoods.

### Results
Squirrel Hill South has the highest number of recorded trees, followed by Highland Park and Squirrel Hill North.

### Conclusion
Based on my individual metric, Squirrel Hill South currently ranks as the best Pittsburgh neighborhood for tree availability.

However, this dataset only represents recorded city-managed trees. It does not include every tree in Pittsburgh, and larger neighborhoods may naturally have more trees. Our group will combine this metric with other outdoor amenities to determine the overall best neighborhood.
